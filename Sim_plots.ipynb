```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
f = "/Users/gabizabi/Downloads/MODEL_airfoil_test.polar"

f1 = "/Users/gabizabi/Downloads/MODEL_airfoil_Ma08.polar"
f2 = "/Users/gabizabi/Downloads/MODEL_airfoil_Ma058.polar"

lines = open(f1).read().splitlines()
i = next(k for k, l in enumerate(lines) if l.split()[:1] == ["Beta"])
pol = pd.read_csv(f, skiprows=i, sep=r"\s+")

lines = open(f2).read().splitlines()
i = next(k for k, l in enumerate(lines) if l.split()[:1] == ["Beta"])
pol2 = pd.read_csv(f, skiprows=i, sep=r"\s+")
```

```python
plt.plot(pol["AoA"], pol["CLtot"], "o-", color = "green", label=f"$C_L$")

plt.grid()
plt.title("Simulation Aircraft Lift Coefficients")
plt.xlabel(f"Angle of attack, α (deg)")
plt.ylabel("Lift Coefficients")
plt.ylim(-.5,2.5)
```

```python
plt.plot(pol["AoA"], pol["CDtot"], "o-", color = "green", label=f"$C_L$")

plt.grid()
plt.title("Simulation Aircraft Drag Coefficients")
plt.xlabel(f"Angle of attack, α (deg)")
plt.ylabel("Drag Coefficients")
plt.ylim(-.5,3.5)
```

```python
plt.plot(pol["AoA"], pol["CMxtot"], "o-", color = "green", label=f"$C_L$")

plt.grid()
plt.title("Simulation Aircraft Moment Coefficients")
plt.xlabel(f"Angle of attack, α (deg)")
plt.ylabel("Moment Coefficients")
plt.ylim(-.1,.1)
```

```python
AR = 2.14
e = 0.7
K = 1/(np.pi * AR * e)

polardrag = 0.028229 + K*pol["CLtot"]**2 

# polar drag = zero lift drag coefficient + K CL**2

plt.plot(polardrag, pol["CLtot"], "-o", color = "green", label=f"$C_L$")

plt.grid()
plt.title("Lift Coefficient vs. Drag Coefficient (Polar)")
plt.ylabel(f"Lift Coefficient")
plt.xlabel("Drag Coefficient (Polar)")
```
